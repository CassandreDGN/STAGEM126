```python
import pandas as pd

df = pd.read_csv("9606.tsv.gz", sep='\t', compression='gzip', 
                 usecols=['protein/inparalogs', 'orthologs', 'ortholog_taxid'])

df_fish = df[df['ortholog_taxid'] == 559292].copy()

#split my list on one line into multiple partd

df_fish['protein/inparalogs'] = df_fish['protein/inparalogs'].str.split(',')
df_fish['orthologs'] = df_fish['orthologs'].str.split(',')

# IA ADDED bc i couldnt figure out how to get each prot on a separate line 
# It turns [ProtA, ProtB] [Orth1, Orth2] into 4 separate rows of pairs.
df_exploded = df_fish.explode('protein/inparalogs').explode('orthologs')

# 5. Save the clean pairs (Col 1: Human, Col 2: Fish)
df_exploded[['protein/inparalogs', 'orthologs']].to_csv("human_yeast_pairs.tsv", 
                                                       sep='\t', 
                                                       header=False, 
                                                       index=False)

print(f" {len(df_exploded)} individual pairs.")
print(df_exploded[['protein/inparalogs', 'orthologs']].head(10))
```
